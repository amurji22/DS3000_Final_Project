```python
import pandas as pd

# Mount Drive
from google.colab import drive
drive.mount('/content/drive')

# load the dataset
df_charging = pd.read_csv('/content/drive/MyDrive/DS 3000/ev_charging_patterns.csv')
df_charging.head()
```

```python
# Overview of dataset columns, types, and missing values
df_charging.info()

# Basic statistics for numerical columns
df_charging.describe()

# Count of missing values per column
df_charging.isnull().sum()

```

```python
# Filling errors/missing values in data-set
# Fill tiny missing 'Energy Consumed (kWh)' with median
df_charging['Energy Consumed (kWh)'].fillna(df_charging['Energy Consumed (kWh)'].median(), inplace=True)

# Fill 'Charging Rate (kW)' and 'Distance Driven (since last charge) (km)' with median
df_charging['Charging Rate (kW)'].fillna(df_charging['Charging Rate (kW)'].median(), inplace=True)
df_charging['Distance Driven (since last charge) (km)'].fillna(df_charging['Distance Driven (since last charge) (km)'].median(), inplace=True)

# Verify missing values were handled
df_charging.isnull().sum()
```

```python
#extract categorical variables
categorical_cols = [
    'Vehicle Model', 'Charging Station Location', 'Time of Day',
    'Day of Week', 'Charger Type', 'User Type'
]

#drop ID columns
drop_cols = ['User ID', 'Charging Station ID']
df_charging.drop(columns=drop_cols, inplace=True, errors='ignore')

# One-hot encode categorical variables
df_encoded = pd.get_dummies(df_charging, columns=categorical_cols, drop_first=True)

df_encoded.head()
```

```python
df_charging['Charging Start Time'] = pd.to_datetime(df_charging['Charging Start Time'])
df_charging['Charging End Time'] = pd.to_datetime(df_charging['Charging End Time'])

# Extract hour of day and weekday as features
df_encoded['start_hour'] = df_charging['Charging Start Time'].dt.hour
df_encoded['start_weekday'] = df_charging['Charging Start Time'].dt.weekday

# Drop original datetime columns if desired
df_encoded.drop(columns=['Charging Start Time', 'Charging End Time'], inplace=True)
```

```python
# Export and Share
# Goal:
#   - Export dataset into organized folders
#   - Ensure teammates can easily find all project files

import os
os.makedirs('data', exist_ok=True)

# df_encoded should already be created in Block-YY
df_encoded.to_csv('data/charging_data_encoded.csv', index=False)

print("  - data/charging_data_encoded.csv")
```
